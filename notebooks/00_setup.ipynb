# 00 · Setup do ambiente
Cria o schema e o volume do projeto no Unity Catalog e copia o dataset bruto do repositório para o volume.

| Objeto | Nome |
|---|---|
| Schema | `workspace.fauna` |
| Volume | `workspace.fauna.raw` |

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.fauna
COMMENT 'Checkpoint - Monitoramento da Fauna do Cerrado';

CREATE VOLUME IF NOT EXISTS workspace.fauna.raw
COMMENT 'Arquivos brutos recebidos';
```

```python
import shutil
from pathlib import Path

origem = Path.cwd().parent / "data" / "dataset.json"
destino = Path("/Volumes/workspace/fauna/raw/dataset.json")

print(f"Origem : {origem} | existe? {origem.exists()}")

shutil.copyfile(origem, destino)
print(f"Copiado para: {destino}")
```
